# **Power BI Visuals**



Dataset: **Global_Superstore**

Common fields used:

* Order Date
* Country / Region / Market
* Category / Sub-Category
* Sales
* Profit
* Quantity
* Customer Name

---

# Step 1: Load the Dataset

1. Open Power BI Desktop.
2. Click **Get Data → Excel/CSV** (depending on your dataset format).
3. Select the Global Superstore dataset.
4. Click **Load**.
5. Ensure these columns are properly formatted:

   * Order Date → Date type
   * Sales → Decimal Number
   * Profit → Decimal Number
   * Quantity → Whole Number

---

# Question 1: Create a Card Visual to Show Total Sales

## Steps

1. Select the **Card Visual** from the Visualizations pane.
2. Drag the **Sales** field into the Values section.
3. Power BI automatically aggregates it as Sum.
4. Rename the title to **Total Sales**.

## DAX Measure (Recommended)

```DAX
Total Sales = SUM('Global_Superstore'[Sales])
```

Use this measure inside the Card visual.

## Insight

The card displays the overall revenue generated from all orders.

---

# Question 2: Create a Donut Chart Showing Total Sales by Ship Mode

## Steps

1. Insert a **Donut Chart**.
2. Drag:

   * Ship Mode → Legend
   * Sales → Values
3. Enable Data Labels.
4. Sort by Sales descending.

## Insight

The ship mode with the largest donut section generates the highest sales.
Usually, in Global Superstore datasets, **Standard Class** generates the highest sales.

---

# Question 3: Create a Line Chart Showing Sales Trend Over Time

## Steps

1. Insert a **Line Chart**.
2. Drag:

   * Order Date → X-Axis
   * Sales → Y-Axis
3. In the date hierarchy, select:

   * Year
   * Month
4. Enable markers and labels if needed.

## Insight

You can identify:

* Seasonal spikes
* Low-performing months
* Year-over-year growth

Typical pattern:

* Higher sales toward year-end (November–December)
* Lower sales during early months

---

# Question 4: Create a Bar Chart to Compare Sales by Category

## Steps

1. Insert a **Clustered Bar Chart**.
2. Drag:

   * Category → Y-Axis
   * Sales → X-Axis
3. Sort descending by Sales.

## Expected Insight

Usually:

* Technology has highest sales
* Furniture second
* Office Supplies third

---

# Question 5: Create a Matrix Visual to Display Sales by Month and Category

## Steps

1. Insert a **Matrix Visual**.
2. Drag:

   * Month (from Order Date) → Rows
   * Category → Columns
   * Sales → Values
3. Apply conditional formatting to highlight highest sales.

## Insight

Identify:

* Best-performing month
* Best-performing category within each month

Typically, November or December records the highest sales.

---

# Question 6: Create a Decomposition Tree for Sales Breakdown

## Steps

1. Insert **Decomposition Tree** visual.
2. Drag:

   * Sales → Analyze
3. Drag these into Explain By:

   * Category
   * Sub-Category
   * Market
   * Ship Mode
   * Region

## Insight

The decomposition tree helps analyze:

* Which category contributes most to sales
* Which market performs best
* Which sub-category drives revenue

# Question 7: Create a Scatter Chart to Compare Discount vs Profit by Sub-Category

## Steps

1. Insert a **Scatter Chart**.
2. Drag:

   * Discount → X-Axis
   * Profit → Y-Axis
   * Sub-Category → Details
   * Sales → Size
3. Enable data labels if required.

## Insight

Look for:

* Points with high discount and negative profit
* Loss-making sub-categories

Typical outliers include:

* Tables
* Bookcases

These often show high discounts with negative profit.

---

# Question 8: Use a Funnel Chart to Show Number of Orders by Customer Segment

## Steps

1. Insert a **Funnel Chart**.
2. Drag:

   * Segment → Group
   * Order ID (Count) → Values

## DAX Measure (Optional)

```DAX
Total Orders = COUNT('Global_Superstore'[Order ID])
```

## Insight

The funnel displays which customer segment contributes the most orders.
Usually:

* Consumer segment has the highest orders
* Corporate second
* Home Office third

---

# Question 9: Use a Waterfall Chart to Analyze Profit Changes by Category within Countries

## Steps

1. Insert a **Waterfall Chart**.
2. Drag:

   * Country → Category
   * Profit → Y-Axis
   * Category → Breakdown

## Insight

The chart shows:

* Profit increase/decrease across categories
* Countries with strong or weak profitability

You can identify categories causing losses in specific countries.

# Question 10: Create a Dual-Axis Line Chart to Show Total Sales and Profit Over Time

## Steps

1. Insert a **Line and Clustered Column Chart**.
2. Drag:

   * Order Date → Shared Axis
   * Sales → Column Values
   * Profit → Line Values
3. Use Month-Year hierarchy.

```DAX
Total Sales = SUM('Global_Superstore'[Sales])
```

```DAX
Total Profit = SUM('Global_Superstore'[Profit])
```
